#1. Importing required libraries

```python
import pandas as pd
import numpy as np
```

#2. Reading the dataset

```python
df=pd.read_csv('dirty_cafe_sales.csv')
df
```



*   Dataset contains 10,000 rows and 8 columns of cafe sale transactions



#3. Identfying data info and description

```python
df.info()
```



*   Most columns contain empty (NaN) values, except for "Transaction ID"
*   "Quantity", "Price Per Unit", and "Total Spent" are being read as text (object) instead of numbers (float)



```python
df.describe()
```

#4. Dropping initial duplicates

```python
#identifying duplicate rows
print(f"Number of duplicate rows before removal: {df.duplicated().sum()}")

#removing duplicate rows
df.drop_duplicates(inplace=True)
print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")

#checking dataframe
df.info()
```

#5. Converting "Quantity", "Price Per Unit", and "Total Spent" into numeric values (float)

#5.1 Converting "Price Per Unit"

```python
#establishing price list
item_price = {
    'Coffee': 2.0,
    'Tea': 1.5,
    'Sandwich': 4.0,
    'Salad': 3.0,
    'Cake': 3.0,
    'Cookie': 1.0,
    'Smoothie': 4.0,
    'Juice': 3.0
}

#reverse mapping price to item
price_item = {price: item for item, price in item_price.items()}

#fill missing 'Item' values based on 'Price Per Unit'
df['Item'] = df['Item'].fillna(df['Price Per Unit'].map(price_item))

#replacing 'ERROR' and 'UNKNOWN' values
df['Price Per Unit'] = df['Price Per Unit'].replace(['ERROR', 'UNKNOWN'], np.nan)

#coverting column into float type
df['Price Per Unit']=df['Price Per Unit'].astype(float)

#checking dataframe
df.info()
```



*   Price for each item is validated (Cookie \$1, Tea \$1.5, Coffee \$2, Salad, Cake, & Juice \$3, and Sandwich & Smoothie \$4)
*   Items listed as 'ERROR' and 'UNKNOWN' were replaced as empty (NaN) values
*   Column has been converted to float type after removing any texts





#5.2 Converting "Total Spent"

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Total Spent'] = df['Total Spent'].replace(['ERROR', 'UNKNOWN'], np.nan)

#coverting column into float type
df['Total Spent']=df['Total Spent'].astype(float)

#checking dataframe
df.info()
```

#5.3 Converting "Quantity"

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Quantity'] = df['Quantity'].replace(['ERROR', 'UNKNOWN'], np.nan)

#coverting column into float type
df['Quantity']=df['Quantity'].astype(float)

#checking dataframe
df.info()
```

#6. Filling empty values in numeric (float) columns

#6.1 Filling "Price Per Unit" empty values

```python
#dividing 'total spent' & 'quantity'
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Total Spent']/df['Quantity'])

#checking dataframe
df.info()
```



*   Divided values from "Total Spent" and "Quantity" to find missing "Price Per Unit" values



```python
#filling 'Item' values from mapped price list
df['Item'] = df['Item'].fillna(df['Price Per Unit'].map(price_item))

#checking dataframe
df.info()
```



*   Values listed as "Item" were replaced by their corresponding "Price Per Unit"



```python
#filling empty 'Price Per Unit' values based on 'Item' name
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Item'].map(item_price))

#checking dataframe
df.info()
```



*   Other missing values are filled based on mapped "Item" prices



#6.2 Filling "Total Spent" empty values

```python
#multiplying 'quantity' & 'price per unit'
df['Total Spent'] = df['Total Spent'].fillna(df['Quantity']*df['Price Per Unit'])

#checking dataframe
df.info()
```



*   Multiplied values from "Quantity" and "Price Per Unit" to find missing "Total Spent" values



#6.3 Filling "Quantity" empty values

```python
#dividing 'total spent' & 'price per unit'
df['Quantity'] = df['Quantity'].fillna(df['Total Spent']/df['Price Per Unit']).round()

#checking dataframe
df.info()
```



*   Divided values from "Total Spent" and "Price Per Unit" to find missing "Quantity" values



#7. Transforming "ERROR", "UNKNOWN", and empty values in text (object) columns

#7.1 Transforming "ERROR", "UNKNOWN", and empty values in "Item" column

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Item'] = df['Item'].replace(['ERROR', 'UNKNOWN'], np.nan)

#identifying $1, $1.5, and $2 items
mask = df['Item'].isna() & ~df['Price Per Unit'].isin([3.0, 4.0])
df.loc[mask, 'Item'] = df.loc[mask, 'Price Per Unit'].map(price_item)

#categorizing empty 'Item' values with 'Price Per Unit' of 3 as '$3 Item'
mask_3_item = df['Item'].isna() & (df['Price Per Unit'] == 3.0)
df.loc[mask_3_item, 'Item'] = '$3 Item'

#categorizing empty 'Item' values with 'Price Per Unit' of 4 as '$4 Item'
mask_4_item = df['Item'].isna() & (df['Price Per Unit'] == 4.0)
df.loc[mask_4_item, 'Item'] = '$4 Item'

#categorizing empty 'Item' values with 'Price Per Unit' of 5 as '$5 Item'
mask_5_item = df['Item'].isna() & (df['Price Per Unit'] == 5.0)
df.loc[mask_5_item, 'Item'] = '$5 Item'

#checking dataframe
df.info()
```



*   Missing items are identified based on their specific "Price Per Unit"
*   Items with "Price Per Unit" of \$3 and \$4 are categorized as "\$3 Item" and "\$4 Item", respectively, due to several items sharing the same "Price Per Unit"
*   Items with "Price Per Unit" of \$5 is categorized as "\$5 Item" due to no specific item name







#7.2 Transforming "ERROR", "UNKNOWN", and empty values in "Payment Method" column

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Payment Method'] = df['Payment Method'].replace(['ERROR', 'UNKNOWN'], np.nan)

#filling empty values with 'Unknown'
df['Payment Method'] = df['Payment Method'].fillna('Unknown')

#checking dataframe
df.info()
```



*   "ERROR", "UNKNOWN", and empty values are categorized to "Unknown"



#7.3 Transforming "ERROR", "UNKNOWN", and empty values in "Location" column

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Location'] = df['Location'].replace(['ERROR', 'UNKNOWN'], np.nan)

#filling empty values with 'Unknown'
df['Location'] = df['Location'].fillna('Unknown')

#checking dataframe
df.info()
```



*   "ERROR", "UNKNOWN", and empty values are categorized to "Unknown"



#7.4 Transforming "ERROR", "UNKNOWN", and empty values in "Transaction Date" column

```python
#replacing 'ERROR' and 'UNKNOWN' values
df['Transaction Date'] = df['Transaction Date'].replace(['ERROR', 'UNKNOWN'], np.nan)

#filling empty values with 'Unknown'
df['Transaction Date'] = df['Transaction Date'].fillna('Unknown')

#checking dataframe
df.info()
```



*   "ERROR", "UNKNOWN", and empty values are categorized to "Unknown"



#8. Removing incomplete records

```python
#removing rows with empty values
df.dropna(subset=['Total Spent'], inplace=True)

#checking dataframe
df.info()
```



*   Removed rows where "Total Spent" was empty, as they provide no financial value
*   Retained all rows with a "Total Spent" value even if other columns are incomplete to ensure the final total revenue remains accurate




#9. Checking for duplicates

```python
#identifying duplicate rows
print(f"Number of duplicate rows before removal: {df.duplicated().sum()}")

#removing duplicate rows
df.drop_duplicates(inplace=True)
print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")

#checking dataframe
df.info()
```
